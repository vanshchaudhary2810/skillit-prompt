```
# One live run of the final chain (v3)
import sys, json
sys.path.insert(0, 'code')
from skillit import chain, validate
from skillit.llm import GeminiClient
client = GeminiClient(min_interval=4)
res = chain.run('v3', {'student_skills': 'python, basic sql, html', 'target_role': 'Data Analyst', 'weekly_hours': 6, 'num_weeks': 4}, client)
print(res['status'], res.get('model'))
print(json.dumps(res.get('final', {}).get('plan'), indent=2))
print(validate.metrics(res))
```
